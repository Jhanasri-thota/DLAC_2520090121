```python
# ============================================================
# VISIONPORT - DEEP LEARNING SKILL LAB SKILL-4
#
# Experiment:
# Project Repository, Git Workflow and
# PyTorch-CUDA Development Environment Setup
#
# Framework : PyTorch
# Model     : Convolutional Neural Network (CNN)
# Tools     : Git, CUDA, Matplotlib
# Platform  : Google Colab
# ============================================================


# ============================================================
# IMPORT REQUIRED LIBRARIES
# ============================================================

# os: Used for directory and file operations.
import os

# sys: Used to access Python interpreter information.
import sys

# time: Used to measure model execution time.
import time

# subprocess: Used to execute Git commands.
import subprocess

# platform: Used to identify the operating system.
import platform

# Path: Provides convenient filesystem path handling.
from pathlib import Path

# PyTorch: Deep learning framework.
import torch

# torch.nn: Contains neural network layers and functions.
import torch.nn as nn

# Matplotlib: Used to generate graphs and visualizations.
import matplotlib.pyplot as plt

# FancyBboxPatch: Used to create diagram boxes.
from matplotlib.patches import FancyBboxPatch


# ============================================================
# 1. CREATE VISIONPORT PROJECT REPOSITORY
# ============================================================

print("=" * 65)
print("1. CREATING VISIONPORT PROJECT REPOSITORY")
print("=" * 65)

# Define the main project directory.
PROJECT_DIR = Path("/content/VisionPort")

# Create the project directory if it does not already exist.
# Existing project files will be preserved.
PROJECT_DIR.mkdir(parents=True, exist_ok=True)

# Define the folders required for the VisionPort project.
folders = [
    "app",
    "models",
    "inference",
    "preprocessing",
    "tests",
    "outputs",
    "docs"
]

# Create each project folder.
for folder in folders:

    # Create the folder inside the main project directory.
    (PROJECT_DIR / folder).mkdir(
        parents=True,
        exist_ok=True
    )

print("Project repository created successfully!")

# Display the project directory structure.
print("\nProject Structure:")

for root, dirs, files in os.walk(PROJECT_DIR):

    # Calculate the directory depth for indentation.
    level = root.replace(str(PROJECT_DIR), "").count(os.sep)

    # Display the current directory name.
    print("  " * level + os.path.basename(root) + "/")


# ============================================================
# 2. CREATE PROJECT FILES
# ============================================================

print("\n" + "=" * 65)
print("2. CREATING PROJECT FILES")
print("=" * 65)


# ------------------------------------------------------------
# CREATE README.md
# ------------------------------------------------------------

# README.md contains the project description and documentation.
readme = """
# VisionPort - Deep Learning Inference Platform

## Description

VisionPort is an end-to-end deep learning inference platform.

## Architecture

Client -> FastAPI -> Preprocessing -> PyTorch Model
-> GPU/CUDA -> Prediction Response

## Technologies

- Python
- PyTorch
- CUDA
- FastAPI
- NVIDIA GPU
- Git and GitHub

## Project Modules

- app: API application
- models: Deep learning models
- inference: Model execution
- preprocessing: Input processing
- tests: Testing modules
- outputs: Results
- docs: Documentation
"""

# Create the README.md file.
(PROJECT_DIR / "README.md").write_text(
    readme.strip() + "\n",
    encoding="utf-8"
)


# ------------------------------------------------------------
# CREATE CNN MODEL FILE
# ------------------------------------------------------------

# Python source code for the VisionPort CNN model.
model_code = '''
import torch
import torch.nn as nn


# Define the VisionPort Convolutional Neural Network.
class VisionPortCNN(nn.Module):

    def __init__(self):

        # Initialize the parent neural network class.
        super().__init__()

        # Feature extraction layers.
        self.features = nn.Sequential(

            # First convolutional layer.
            nn.Conv2d(
                3,
                16,
                kernel_size=3,
                padding=1
            ),

            # ReLU introduces non-linearity.
            nn.ReLU(),

            # Max pooling reduces spatial dimensions.
            nn.MaxPool2d(2),

            # Second convolutional layer.
            nn.Conv2d(
                16,
                32,
                kernel_size=3,
                padding=1
            ),

            # Apply ReLU activation.
            nn.ReLU(),

            # Reduce spatial dimensions again.
            nn.MaxPool2d(2),

            # Convert feature maps into 1x1 dimensions.
            nn.AdaptiveAvgPool2d((1, 1))
        )

        # Classification layer produces 10 output classes.
        self.classifier = nn.Linear(32, 10)

    def forward(self, x):

        # Extract image features.
        x = self.features(x)

        # Flatten feature maps into a feature vector.
        x = torch.flatten(x, 1)

        # Generate classification scores.
        return self.classifier(x)
'''

# Create the model.py file.
(PROJECT_DIR / "models" / "model.py").write_text(
    model_code.strip() + "\n",
    encoding="utf-8"
)


# ------------------------------------------------------------
# CREATE requirements.txt
# ------------------------------------------------------------

# Store the required project dependencies.
requirements = """
torch
matplotlib
fastapi
"""

(PROJECT_DIR / "requirements.txt").write_text(
    requirements.strip() + "\n",
    encoding="utf-8"
)


# ------------------------------------------------------------
# CREATE .gitignore
# ------------------------------------------------------------

# Exclude unnecessary files from Git tracking.
gitignore = """
__pycache__/
*.pyc
.env
.ipynb_checkpoints/
outputs/*.pt
outputs/*.pth
"""

(PROJECT_DIR / ".gitignore").write_text(
    gitignore.strip() + "\n",
    encoding="utf-8"
)

print("Project files created successfully!")

# Display all created files.
print("\nProject Files:")

for file in PROJECT_DIR.rglob("*"):

    if file.is_file():

        print(" -", file.relative_to(PROJECT_DIR))


# ============================================================
# 3. GIT INITIALIZATION AND VERSION CONTROL WORKFLOW
# ============================================================

print("\n" + "=" * 65)
print("3. GIT VERSION CONTROL WORKFLOW")
print("=" * 65)


# ------------------------------------------------------------
# DEFINE GIT COMMAND FUNCTION
# ------------------------------------------------------------

# This function executes Git commands inside the project folder.
def git(command):

    result = subprocess.run(
        ["git"] + command,
        cwd=PROJECT_DIR,
        capture_output=True,
        text=True
    )

    # Raise an error if a Git command fails.
    if result.returncode != 0:

        raise RuntimeError(
            f"Git command failed: {' '.join(command)}\n"
            f"{result.stderr}"
        )

    return result.stdout.strip()


# ------------------------------------------------------------
# INITIALIZE GIT REPOSITORY
# ------------------------------------------------------------

# Initialize Git only if the repository does not already exist.
if not (PROJECT_DIR / ".git").exists():

    git(["init", "-b", "main"])

    print("Git repository initialized successfully!")

else:

    print("Git repository already exists.")


# ------------------------------------------------------------
# CONFIGURE GIT USER
# ------------------------------------------------------------

# Configure the Git username after repository initialization.
git([
    "config",
    "user.name",
    "VisionPort Student"
])

# Configure the Git email after repository initialization.
git([
    "config",
    "user.email",
    "student@example.com"
])

print("Git username and email configured successfully!")


# ------------------------------------------------------------
# ENSURE MAIN BRANCH EXISTS / IS SELECTED
# ------------------------------------------------------------

has_commits = False
try:
    # A safer check for existing commits is to verify if HEAD points to a valid commit.
    git(["show-ref", "--head", "HEAD"])
    has_commits = True
except RuntimeError:
    pass

if has_commits:
    branches = git(["branch", "--list", "main"])
    if not branches:
        git(["branch", "-M", "main"])
    git(["switch", "main"])
    print("Main branch selected successfully!")
else:
    print("Initial repository has no commits yet. Main branch is active by default.")


# ------------------------------------------------------------
# CREATE INITIAL COMMIT
# ------------------------------------------------------------

# Add all project files to the Git staging area.
git(["add", "."])

# Check whether there are staged changes.
staged_changes = git([
    "diff",
    "--cached",
    "--name-only"
])

# Create the initial commit only when staged changes exist.
if staged_changes:

    git([
        "commit",
        "-m",
        "Initial VisionPort project setup"
    ])

    print("\nInitial commit created successfully!")

else:

    print("\nNo new files to commit. Repository is up to date.")


# Display Git commit history.
print("\nInitial Git History:")

try:
    print(git([
        "log",
        "--oneline",
        "-5"
    ]))
except RuntimeError:
    print("No history available yet (no commits).")


# ------------------------------------------------------------
# CREATE FEATURE BRANCH
# ------------------------------------------------------------

feature_branch = "feature/environment-setup"

# Check whether the feature branch already exists.
existing_branches = git([
    "branch",
    "--list",
    feature_branch
])

if existing_branches:

    # Switch to the existing feature branch.
    git([
        "switch",
        feature_branch
    ])

    print("\nExisting feature branch selected.")

else:

    # Create and switch to the feature branch.
    git([
        "switch",
        "-c",
        feature_branch
    ])

    print("\nFeature branch created successfully!")


# ------------------------------------------------------------
# CREATE ENVIRONMENT DOCUMENTATION
# ------------------------------------------------------------

# Create environment documentation.
environment_info = """
VisionPort Development Environment

Framework: PyTorch
Hardware: CPU / CUDA GPU
Version Control: Git
Platform: Google Colab
Model: Convolutional Neural Network
"""

(PROJECT_DIR / "docs" / "environment.txt").write_text(
    environment_info.strip() + "\n",
    encoding="utf-8"
)

print("Environment documentation created successfully!")


# ------------------------------------------------------------
# COMMIT FEATURE BRANCH CHANGES
# ------------------------------------------------------------

# Add documentation to the Git staging area.
git(["add", "."])

# Display staged file changes.
print("\nStaged Changes:")

print(git([
    "diff",
    "--cached",
    "--stat"
]))

# Commit only if there are staged changes.
staged_changes = git([
    "diff",
    "--cached",
    "--name-only"
])

if staged_changes:

    git([
        "commit",
        "-m",
        "Add development environment documentation"
    ])

    print("Feature branch changes committed successfully!")

else:

    print("No new feature branch changes to commit.")


# ------------------------------------------------------------
# MERGE FEATURE BRANCH INTO MAIN
# ------------------------------------------------------------

# Switch back to the main branch.
git(["switch", "main"])

print("\nSwitched back to main branch.")

# Merge the feature branch into main.
# If already merged, Git reports that the branch is up to date.
merge_output = git([
    "merge",
    "--no-ff",
    feature_branch,
    "-m",
    "Merge environment setup"
])

print("\nMerge Result:")

print(merge_output if merge_output else "Merge completed successfully!")


# ------------------------------------------------------------
# DISPLAY FINAL GIT INFORMATION
# ------------------------------------------------------------

# Display all available branches.
print("\nFinal Git Branches:")

print(git(["branch"]))

# Display the complete Git commit history.
print("\nFinal Git History:")

print(git([
    "log",
    "--oneline",
    "--graph",
    "--all"
]))


# ============================================================
# 4. PYTORCH AND CUDA ENVIRONMENT VERIFICATION
# ============================================================

print("\n" + "=" * 65)
print("4. PYTORCH AND CUDA ENVIRONMENT")
print("=" * 65)


# Display the installed Python version.
print(
    "Python Version:",
    sys.version.split()[0]
)

# Display the operating system.
print(
    "Operating System:",
    platform.system()
)

# Display the installed PyTorch version.
print(
    "PyTorch Version:",
    torch.__version__
)

# Check whether CUDA is available.
gpu_available = torch.cuda.is_available()

print(
    "CUDA Available:",
    gpu_available
)


# ------------------------------------------------------------
# SELECT EXECUTION DEVICE
# ------------------------------------------------------------

if gpu_available:

    # Use NVIDIA GPU when CUDA is available.
    device = torch.device("cuda")

    # Display GPU information.
    print(
        "GPU Name:",
        torch.cuda.get_device_name(0)
    )

    # Display CUDA version.
    print(
        "CUDA Version:",
        torch.version.cuda
    )

    # Display the number of available GPUs.
    print(
        "GPU Count:",
        torch.cuda.device_count()
    )

    # Calculate GPU memory in GB.
    gpu_memory = (
        torch.cuda.get_device_properties(0).total_memory
        / (1024 ** 3)
    )

    print(
        "GPU Memory:",
        round(gpu_memory, 2),
        "GB"
    )

else:

    # Use CPU when CUDA is unavailable.
    device = torch.device("cpu")

    print("GPU not available. Running using CPU.")


# Display the selected execution device.
print(
    "Selected Device:",
    device
)


# ============================================================
# 5. CREATE AND EXECUTE CNN DEEP LEARNING MODEL
# ============================================================

print("\n" + "=" * 65)
print("5. CNN MODEL EXECUTION")
print("=" * 65)


# ------------------------------------------------------------
# DEFINE VISIONPORT CNN MODEL
# ------------------------------------------------------------

class VisionPortCNN(nn.Module):

    def __init__(self):

        # Initialize the parent neural network class.
        super().__init__()

        # Feature extraction layers.
        self.features = nn.Sequential(

            # First convolutional layer.
            nn.Conv2d(3, 16, kernel_size=3, padding=1),

            # ReLU activation function.
            nn.ReLU(),

            # Max pooling layer.
            nn.MaxPool2d(2),

            # Second convolutional layer.
            nn.Conv2d(16, 32, kernel_size=3, padding=1),

            # ReLU activation.
            nn.ReLU(),

            # Second max pooling layer.
            nn.MaxPool2d(2),

            # Global average pooling.
            nn.AdaptiveAvgPool2d((1, 1))
        )

        # Fully connected classification layer.
        self.classifier = nn.Linear(32, 10)

    def forward(self, x):

        # Extract image features.
        x = self.features(x)

        # Flatten the extracted features.
        x = torch.flatten(x, 1)

        # Generate classification scores.
        return self.classifier(x)


# ------------------------------------------------------------
# INITIALIZE MODEL
# ------------------------------------------------------------

# Create the CNN model.
model = VisionPortCNN()

# Move the model to the selected device.
model = model.to(device)

# Set the model to evaluation mode.
model.eval()

print("VisionPort CNN model created successfully!")

# Display the model architecture.
print("\nModel Architecture:")

print(model)


# ------------------------------------------------------------
# CREATE SAMPLE INPUT DATA
# ------------------------------------------------------------

# Generate 4 sample RGB images.
# Each image has 3 channels and dimensions 64x64.
sample_input = torch.randn(
    4,
    3,
    64,
    64
)

# Move input data to the selected device.
sample_input = sample_input.to(device)

print("\nInput Tensor Shape:")

print(sample_input.shape)


# ------------------------------------------------------------
# EXECUTE CNN MODEL
# ------------------------------------------------------------

# Disable gradient calculation during inference.
with torch.no_grad():

    # Execute the CNN model.
    prediction = model(sample_input)


# Display prediction tensor shape.
print("\nPrediction Tensor Shape:")

print(prediction.shape)

# Display the number of model parameters.
print(
    "\nNumber of Parameters:",
    sum(
        p.numel()
        for p in model.parameters()
    )
)

# Find the predicted class for each image.
predicted_classes = torch.argmax(
    prediction,
    dim=1
)

# Convert predictions into a Python list.
print(
    "Predicted Classes:",
    predicted_classes.cpu().tolist()
)


# ------------------------------------------------------------
# SAVE MODEL WEIGHTS
# ------------------------------------------------------------

# Save the model state dictionary.
# Note: This model has randomly initialized weights.
# It is not trained on a real dataset.
model_path = (
    PROJECT_DIR
    / "outputs"
    / "visionport_model.pth"
)

torch.save(
    model.state_dict(),
    model_path
)

print("\nModel saved successfully!")

print("Model Path:", model_path)


# ============================================================
# 6. CPU VS GPU PERFORMANCE BENCHMARK
# ============================================================

print("\n" + "=" * 65)
print("6. CPU VS GPU PERFORMANCE BENCHMARK")
print("=" * 65)


# ------------------------------------------------------------
# DEFINE BENCHMARK FUNCTION
# ------------------------------------------------------------

def benchmark(device, iterations=20):

    # Create a separate CNN model for benchmarking.
    benchmark_model = VisionPortCNN().to(device)

    # Set model to evaluation mode.
    benchmark_model.eval()

    # Generate sample image data.
    data = torch.randn(
        8,
        3,
        64,
        64
    ).to(device)

    # Disable gradient calculations.
    with torch.no_grad():

        # Perform warm-up iterations.
        for _ in range(5):

            benchmark_model(data)

        # Synchronize CUDA operations before timing.
        if device.type == "cuda":

            torch.cuda.synchronize()

        # Record the starting time.
        start = time.perf_counter()

        # Execute the model multiple times.
        for _ in range(iterations):

            benchmark_model(data)

        # Wait for GPU operations to finish.
        if device.type == "cuda":

            torch.cuda.synchronize()

        # Record the ending time.
        end = time.perf_counter()

    # Calculate average execution time in milliseconds.
    average_time = (
        (end - start) / iterations
    ) * 1000

    return average_time


# ------------------------------------------------------------
# BENCHMARK CPU
# ------------------------------------------------------------

cpu_time = benchmark(
    torch.device("cpu")
)

# Store CPU timing results.
times = {
    "CPU": cpu_time
}

print(
    f"CPU Average Time: {cpu_time:.4f} ms"
)


# ------------------------------------------------------------
# BENCHMARK CUDA GPU
# ------------------------------------------------------------

if gpu_available:

    # Benchmark using CUDA GPU.
    gpu_time = benchmark(
        torch.device("cuda")
    )

    # Store GPU timing results.
    times["CUDA GPU"] = gpu_time

    print(
        f"GPU Average Time: {gpu_time:.4f} ms"
    )

    # Calculate speedup.
    if gpu_time > 0:

        speedup = cpu_time / gpu_time

        print(
            "Measured Speedup:",
            f"{speedup:.2f}x"
        )

else:

    print(
        "CUDA benchmark skipped because GPU is unavailable."
    )


# ============================================================
# 7. GRAPHICAL OUTPUT 1 - VISIONPORT ARCHITECTURE
# ============================================================

print("\n" + "=" * 65)
print("7. GRAPHICAL OUTPUTS")
print("=" * 65)


# Create a figure for the architecture diagram.
fig, ax = plt.subplots(figsize=(15, 5))

# Define diagram boundaries.
ax.set_xlim(0, 15)
ax.set_ylim(0, 5)

# Hide graph axes.
ax.axis("off")


# Define system architecture components and positions.
architecture = [
    ("Client", 0.3),
    ("FastAPI", 2.7),
    ("Preprocessing", 5.1),
    ("PyTorch Model", 7.5),
    ("GPU / CUDA", 10.0),
    ("Prediction", 12.5)
]


# Draw each architecture component.
for label, x in architecture:

    # Create a rounded rectangular box.
    box = FancyBboxPatch(
        (x, 2),
        2,
        1.1,
        boxstyle="round,pad=0.12",
        linewidth=1.8,
        edgecolor="navy",
        facecolor="lightblue"
    )

    # Add the box to the diagram.
    ax.add_patch(box)

    # Add the component name inside the box.
    ax.text(
        x + 1,
        2.55,
        label,
        ha="center",
        va="center",
        fontsize=10,
        fontweight="bold"
    )


# Draw arrows connecting components.
for i in range(len(architecture) - 1):

    x1 = architecture[i][1] + 2
    x2 = architecture[i + 1][1]

    ax.annotate(
        "",
        xy=(x2 - 0.1, 2.55),
        xytext=(x1 + 0.1, 2.55),
        arrowprops=dict(
            arrowstyle="->",
            lw=2
        )
    )


# Add the diagram title.
ax.text(
    7.5,
    4.2,
    "VISIONPORT END-TO-END DEEP LEARNING ARCHITECTURE",
    ha="center",
    fontsize=14,
    fontweight="bold"
)

# Add the workflow description.
ax.text(
    7.5,
    0.8,
    "Input Request -> Processing -> Inference -> Prediction Response",
    ha="center",
    fontsize=11
)

# Adjust layout and display the diagram.
plt.tight_layout()
plt.show()


# ============================================================
# 8. GRAPHICAL OUTPUT 2 - PERFORMANCE COMPARISON
# ============================================================

# Create a performance comparison chart.
plt.figure(figsize=(9, 5))

# Draw bars for the available execution devices.
bars = plt.bar(
    list(times.keys()),
    list(times.values())
)

# Add chart title and axis labels.
plt.title(
    "VisionPort: CPU vs CUDA GPU Inference Time"
)

plt.xlabel("Execution Device")

plt.ylabel(
    "Average Execution Time (milliseconds)"
)

# Display horizontal grid lines.
plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.5
)

# Display execution-time values above each bar.
for bar, value in zip(bars, times.values()):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value:.3f} ms",
        ha="center",
        va="bottom"
    )

# Adjust layout and display the chart.
plt.tight_layout()
plt.show()


# ============================================================
# 9. GRAPHICAL OUTPUT 3 - GIT WORKFLOW
# ============================================================

# Create a figure for the Git workflow diagram.
fig, ax = plt.subplots(figsize=(11, 4))

# Set diagram boundaries.
ax.set_xlim(0, 11)
ax.set_ylim(0, 4)

# Hide graph axes.
ax.axis("off")


# Define Git workflow stages.
git_steps = [
    ("Working\nDirectory", 0.5),
    ("Staging\nArea", 3.0),
    ("Local\nRepository", 5.5),
    ("Feature\nBranch", 8.0)
]


# Draw the workflow boxes.
for label, x in git_steps:

    box = FancyBboxPatch(
        (x, 1.5),
        2,
        1,
        boxstyle="round,pad=0.1",
        facecolor="lightgreen",
        edgecolor="darkgreen",
        linewidth=2
    )

    # Add each box to the figure.
    ax.add_patch(box)

    # Add the stage name.
    ax.text(
        x + 1,
        2,
        label,
        ha="center",
        va="center",
        fontsize=11,
        fontweight="bold"
    )


# Draw arrows between Git workflow stages.
for i in range(3):

    ax.annotate(
        "",
        xy=(git_steps[i + 1][1] - 0.1, 2),
        xytext=(git_steps[i][1] + 2.1, 2),
        arrowprops=dict(
            arrowstyle="->",
            lw=2
        )
    )


# Add the Git workflow title.
ax.text(
    5.5,
    3.4,
    "GIT VERSION CONTROL WORKFLOW",
    ha="center",
    fontsize=14,
    fontweight="bold"
)

# Display important Git commands.
ax.text(
    5.5,
    0.7,
    "git add -> git commit -> git branch -> git merge",
    ha="center",
    fontsize=11
)

# Adjust layout and display the diagram.
plt.tight_layout()
plt.show()


# ============================================================
# 10. FINAL EXPERIMENT SUMMARY
# ============================================================

print("\n" + "=" * 65)
print("EXPERIMENT COMPLETED SUCCESSFULLY")
print("=" * 65)


# Display completed experiment tasks.
print("1. VisionPort repository created.")

print("2. Project folders and files created.")

print("3. Git initialization and commits completed.")

print("4. Feature branch created and merged.")

print("5. PyTorch environment verified.")

print("6. CUDA/GPU availability checked.")

print("7. CNN model executed successfully.")

print("8. Model weights saved.")

print("9. CPU/GPU performance measured.")

print("10. Graphical outputs generated.")


# Display the project repository location.
print("\nRepository Location:")

print(PROJECT_DIR)


# Display the current Git branch.
print(
    "\nCurrent Git Branch:",
    git(["branch", "--show-current"])
)


# Display final Git status.
print("\nGit Status:")

status = git(["status", "--short"])

if status:

    print(status)

else:

    print("Working tree clean.")


# Display final completion message.
print("\nALL LAB TASKS FINISHED!")

print("=" * 65)

# ============================================================
# END OF VISIONPORT SKILL-4 EXPERIMENT
# ============================================================
```
